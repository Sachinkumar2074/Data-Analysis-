```python
import pandas as pd
import streamlit as st
```

```python
df = pd.read_excel("../dataset/Financial_Sample.xlsx")
```

```python
df.shape
```

```python
import seaborn as sns
```

```python
sns.heatmap(df.isnull())
```

```python
for col in df.columns:
    if df[col].isnull().sum()>0:
        if df[col].dtype == float:
            df[col] = df[col].fillna(round(df[col].mean(),2))
        else:
            df[col] = df[col].fillna(df[col].mode()[0])
```

```python
sns.heatmap(df.isnull())
```

```python
df['GrossManu'] = df['Units Sold']*df['Manufacturing Price']
```

```python
df.to_csv("../dataset/cleaned.csv", index=False)
```
